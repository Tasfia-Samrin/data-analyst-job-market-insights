```python
import ast
from collections import Counter

# Convert description_tokens from text into actual Python lists
df['skills'] = df['description_tokens'].apply(ast.literal_eval)

# Count how many job postings mention each skill
skill_counts = Counter(
    skill
    for skills in df['skills'] #goes through the skills column one row at a time.
    for skill in skills #going inside each individual list.
)

# Show the 30 most requested skills
print("TOP 30 SKILLS")
print(skill_counts.most_common(30))

# Number of postings with at least one identified skill
print("\nPOSTINGS WITH SKILLS")
print((df['skills'].str.len() > 0).value_counts())

# Average number of identified skills per posting
print("\nAVERAGE SKILLS PER POSTING")
print(df['skills'].str.len().mean())
```

```python
print("JUNIOR TITLES")
print(
    df[df['title'].str.contains('Junior', case=False, na=False)]['title']
    .value_counts()
    .head(20)
)

print("\nSENIOR TITLES")
print(
    df[df['title'].str.contains(r'Senior|Sr\.', case=False, na=False, regex=True)]['title']
    .value_counts()
    .head(20)
)

print("\nLEAD TITLES")
print(
    df[df['title'].str.contains('Lead', case=False, na=False)]['title']
    .value_counts()
    .head(20)
)
```

```python
import numpy as np

df['seniority'] = np.select(
    [
        df['title'].str.contains('Lead', case=False, na=False),
        df['title'].str.contains(r'Senior|Sr\.?', case=False, na=False, regex=True),
        df['title'].str.contains('Junior', case=False, na=False)
    ],
    [
        'Lead',
        'Senior',
        'Junior'
    ],
    default='Other'
)

print("SENIORITY DISTRIBUTION")
print(df['seniority'].value_counts())

print("\nSENIORITY PERCENTAGE")
print((df['seniority'].value_counts(normalize=True) * 100).round(2))
```

```python
print("TOP SKILLS BY SENIORITY")

for level in ['Junior', 'Senior', 'Lead']:
    print(f"\n--- {level} ---")
    
    level_skills = Counter(
        skill
        for skills in df.loc[df['seniority'] == level, 'skills']
        for skill in skills
    )
    
    print(level_skills.most_common(10))
```

```python
skills_to_compare = ['sql', 'excel', 'python', 'power_bi', 'tableau', 'r',
                     'sas', 'azure', 'aws', 'snowflake']

skill_by_seniority = {}

for level in ['Junior', 'Senior', 'Lead']:
    level_data = df[df['seniority'] == level]
    total_jobs = len(level_data)

    skill_by_seniority[level] = {}

    for skill in skills_to_compare:
        count = sum(skill in skills for skills in level_data['skills'])
        percentage = (count / total_jobs) * 100
        skill_by_seniority[level][skill] = round(percentage, 2)

skill_by_seniority
```

```python
skill_seniority_df = pd.DataFrame(skill_by_seniority).T # converts to table and .T=transporse the table

skill_seniority_df
```

```python
print("SALARY AVAILABILITY")
print(df['salary'].notna().value_counts())
print((df['salary'].notna().mean() * 100).round(2))

print("\nSALARY RATE")
print(df['salary_rate'].value_counts(dropna=False))

print("\nSTANDARDIZED SALARY")
print(df['salary_standardized'].describe())

print("\nSALARY BY SENIORITY")
print(
    df.groupby('seniority')['salary_standardized']
      .agg(['count', 'mean', 'median', 'min', 'max'])
      .round(2)
)
```

```python
salary_q1 = df['salary_standardized'].quantile(0.25)
salary_q3 = df['salary_standardized'].quantile(0.75)
salary_iqr = salary_q3 - salary_q1

upper_limit = salary_q3 + 1.5 * salary_iqr

print("Q1:", salary_q1)
print("Q3:", salary_q3)
print("IQR:", salary_iqr)
print("Upper outlier limit:", round(upper_limit, 2))

print("\nPotential high-salary outliers:")
print(
    df[df['salary_standardized'] > upper_limit]
    [['title', 'company_name', 'salary', 'salary_rate', 'salary_standardized']]
    .sort_values('salary_standardized', ascending=False)
    .head(20)
)
```

```python
salary_clean = df[
    df['salary_standardized'].notna() &
    (df['salary_standardized'] <= upper_limit)
].copy()

print("Original salary records:", df['salary_standardized'].notna().sum())
print("Clean salary records:", len(salary_clean))
print("Removed as high-salary outliers:",
      df['salary_standardized'].notna().sum() - len(salary_clean))

print("\nCLEAN SALARY SUMMARY")
print(
    salary_clean['salary_standardized']
    .describe()
    .round(2)
)

print("\nCLEAN SALARY BY SENIORITY")
print(
    salary_clean.groupby('seniority')['salary_standardized']
    .agg(['count', 'mean', 'median'])
    .round(2)
)
```

```python
print("UNIQUE LOCATIONS:", df['location'].nunique())

print("\nTOP LOCATIONS")
print(df['location'].value_counts(dropna=False).head(30))

print("\nMISSING LOCATIONS:", df['location'].isna().sum())
```

```python
df['location_clean'] = df['location'].str.strip()

print("Original unique locations:", df['location'].nunique())
print("Cleaned unique locations:", df['location_clean'].nunique())

print("\nTOP CLEANED LOCATIONS")
print(df['location_clean'].value_counts(dropna=False).head(30))
```

```python
df['location_type'] = df['location_clean'].replace({
    'Anywhere': 'Anywhere / Remote',
    'United States': 'United States / Unspecified'
})

print("LOCATION TYPE")
print(df['location_type'].value_counts().head(10))

print("\nTOP SPECIFIC LOCATIONS")
specific_locations = df[
    ~df['location_clean'].isin(['Anywhere', 'United States'])
]

print(
    specific_locations['location_clean']
    .value_counts()
    .head(20)
)
```

```python
# Create the Jobs table

jobs_df = df[
    [
        'job_id',
        'title',
        'company_name',
        'location_clean',
        'location_type',
        'seniority',
        'schedule_type',
        'work_from_home',
        'salary_standardized'
    ]
].copy()

# Rename columns to make them easier to understand
jobs_df = jobs_df.rename(columns={
    'location_clean': 'location',
    'salary_standardized': 'salary_annual'
})


#  Create the Skills table

skills_df = df[
    ['job_id', 'skills']
].explode('skills')

skills_df = skills_df.rename(columns={
    'skills': 'skill'
})

# Remove rows where no skill was identified
skills_df = skills_df.dropna(subset=['skill'])

# Remove duplicate job-skill combinations
skills_df = skills_df.drop_duplicates()


# Check the final tables

print("JOBS TABLE")
print("Rows:", len(jobs_df))
print("Columns:", jobs_df.shape[1])
print(jobs_df.head())

print("\nSKILLS TABLE")
print("Rows:", len(skills_df))
print("Columns:", skills_df.shape[1])
print(skills_df.head())
```

```python
print("Unique jobs:", jobs_df['job_id'].nunique())
print("Unique job IDs in skills table:", skills_df['job_id'].nunique())

missing_job_ids = ~skills_df['job_id'].isin(jobs_df['job_id'])

print("Skills linked to missing jobs:", missing_job_ids.sum())

print("\nJobs with at least one identified skill:",
      skills_df['job_id'].nunique())

print("Jobs without identified skills:",
      len(jobs_df) - skills_df['job_id'].nunique())
```

```python
duplicate_job_ids = jobs_df[
    jobs_df['job_id'].duplicated(keep=False)
].sort_values('job_id')

print("Rows involved in duplicate job IDs:",
      len(duplicate_job_ids))

print("Number of duplicated job IDs:",
      duplicate_job_ids['job_id'].nunique())

print("\nExample duplicated job IDs:")
print(
    duplicate_job_ids[
        ['job_id', 'title', 'company_name', 'location', 'schedule_type']
    ].head(20)
)
```

```python
jobs_df.to_csv("jobs_final.csv", index=False)
skills_df.to_csv("skills_final.csv", index=False)

print("Export complete.")
print("Jobs:", len(jobs_df))
print("Skills:", len(skills_df))
```
